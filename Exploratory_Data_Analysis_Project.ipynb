```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = {
    "Study_Hours": [2, 3, 4, 5, 6, 7, 8, 3, 5, 6],
    "Attendance": [60, 65, 70, 75, 80, 85, 90, 68, 78, 88],
    "Marks": [45, 50, 55, 65, 70, 80, 90, 52, 68, 85]
}

df = pd.DataFrame(data)

print("First five rows:")
display(df.head())

print("Dataset shape:", df.shape)

print("Dataset information:")
df.info()

print("Statistical summary:")
display(df.describe())

print("Missing values:")
print(df.isnull().sum())
```

```python
# Study Hours vs Marks
plt.figure(figsize=(6, 4))
plt.scatter(df["Study_Hours"], df["Marks"])
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Study Hours vs Marks")
plt.show()

# Attendance vs Marks
plt.figure(figsize=(6, 4))
plt.scatter(df["Attendance"], df["Marks"])
plt.xlabel("Attendance")
plt.ylabel("Marks")
plt.title("Attendance vs Marks")
plt.show()

# Correlation Heatmap
plt.figure(figsize=(6, 4))
sns.heatmap(df.corr(), annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

```python
# Correlation values
print("Correlation with Marks:")
print(df.corr()["Marks"].sort_values(ascending=False))

# Highest and lowest marks
print("\nHighest Marks:", df["Marks"].max())
print("Lowest Marks:", df["Marks"].min())
print("Average Marks:", df["Marks"].mean())

# Highest study hours
print("\nMaximum Study Hours:", df["Study_Hours"].max())

# Highest attendance
print("Maximum Attendance:", df["Attendance"].max())

# Check duplicate rows
print("\nDuplicate Rows:", df.duplicated().sum())

# Check missing values
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
print("EXPLORATORY DATA ANALYSIS REPORT")
print("=" * 40)

print("\n1. Dataset Overview")
print("Total Rows:", df.shape[0])
print("Total Columns:", df.shape[1])

print("\n2. Statistical Findings")
print("Average Marks:", round(df["Marks"].mean(), 2))
print("Highest Marks:", df["Marks"].max())
print("Lowest Marks:", df["Marks"].min())

print("\n3. Data Quality")
print("Missing Values:", df.isnull().sum().sum())
print("Duplicate Rows:", df.duplicated().sum())

print("\n4. Key Insights")
print("Study Hours and Marks Correlation:",
      round(df["Study_Hours"].corr(df["Marks"]), 2))
print("Attendance and Marks Correlation:",
      round(df["Attendance"].corr(df["Marks"]), 2))

print("\n5. Conclusion")
print("The dataset shows the relationship between")
print("study hours, attendance, and student marks.")
print("Visualizations and correlation analysis help")
print("identify patterns in the sample data.")
```
